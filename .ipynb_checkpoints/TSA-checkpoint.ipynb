## 1. Uploading Dataset

**Dataset:** 'Hourly Energy Consumption' link: https://www.kaggle.com/datasets/robikscube/hourly-energy-consumption/data

**Brief:** Over 10 years of hourly energy consumption data from PJM in Megawatts. 

**Characterisitcs:** PJM Hourly Energy Consumption Data.

PJM Interconnection LLC (PJM) is a regional transmission organization (RTO) in the United States. It is part of the Eastern Interconnection grid operating an electric transmission system serving all or parts of Delaware, Illinois, Indiana, Kentucky, Maryland, Michigan, New Jersey, North Carolina, Ohio, Pennsylvania, Tennessee, Virginia, West Virginia, and the District of Columbia.

The hourly power consumption data comes from PJM's website and are in megawatts (MW).

The regions have changed over the years so data may only appear for certain dates per region.

## 2. Data Visualisation

In this section, we carry out the dataset visualization which is a critical step in time series analysis. It helps us understand the nature of the data and identify key patterns or characteristics.

```python
# Timeserie distribution
df_aep.plot(
        figsize=(15, 5),
        title='AEP Energy Use in MW')
plt.show()
```

```python
sns.histplot(data = df_aep, x = 'AEP_MW', kde=True, bins = 150)
itle='AEP Energy Use Distirbution'
plt.xlabel('Energy Usage')
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
# Decomposition
seasonal_decompose(df_aep, period=365).plot()
plt.show()
```

```python
# Features Creating

df_aep['hour']=df_aep.index.hour
df_aep['dayOfWeek']=df_aep.index.dayofweek
df_aep['month']=df_aep.index.month
df_aep['year']=df_aep.index.year

def get_season(month):
    if month in [12, 1, 2]:
        return 3
    if month in [3, 4, 5]:
        return 0
    if month in [6, 7, 8]:
        return 1
    if month in [9, 10, 11]:
        return 2

df_aep['season'] = df_aep['month'].apply(get_season)

# 0 - spring
# 1 - summer
# 2 - autumn
# 3 - winter
```

```python
df_aep.head(5)
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
sns.boxplot(data=df_aep, x='hour', y='AEP_MW')
ax.set_title('Energy usage by Hour')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
sns.boxplot(data=df_aep, x='month', y='AEP_MW')
ax.set_title('Energy usage by Hour')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
sns.boxplot(data=df_aep, x='year', y='AEP_MW')
ax.set_title('Energy usage by Hour')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
sns.boxplot(data=df_aep, x='season', y='AEP_MW')
ax.set_title('Energy usage by Hour')
plt.show()
```

```python
index_season = df_aep.groupby('season')['AEP_MW'].mean().sort_values(ascending=False).index
sns.barplot(
    data = df_aep,
    x = 'season',
    y = 'AEP_MW',
    order = index_season,
    errorbar = None,
    estimator=np.mean
)
```

```python
index_season = df_aep.groupby('month')['AEP_MW'].mean().sort_values(ascending=False).index
sns.barplot(
    data = df_aep,
    x = 'month',
    y = 'AEP_MW',
    order = index_season,
    errorbar = None,
    estimator=np.mean
)
```

```python
index_season = df_aep.groupby('year')['AEP_MW'].mean().sort_values(ascending=False).index
sns.barplot(
    data = df_aep,
    x = 'year',
    y = 'AEP_MW',
    order = index_season,
    errorbar = None,
    estimator=np.mean
)
```

```python
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.graphics.tsaplots import plot_pacf
```

```python
f, ax = plt.subplots(nrows=2, ncols=1, figsize=(16, 8))

plot_acf(df_aep['AEP_MW'], lags=100, ax=ax[0])
plot_pacf(df_aep['AEP_MW'], lags=100, ax=ax[1])

plt.show()
```

```python
# Train Test Split
df_train, df_test = df_aep[df_aep.index < '2016-01-01'], df_aep[df_aep.index >= '2016-01-01']

# Calculate the sizes of the training and testing datasets
train_size = len(df_train)
test_size = len(df_test)
total_size = train_size + test_size

# Calculate percentages
train_percentage = (train_size / total_size) * 100
test_percentage = (test_size / total_size) * 100

print('Train:\t', len(df_train))
print('Test:\t', len(df_test))

train_percentage, test_percentage
```

```python
plt.figure(figsize=(20,8))

df_train['AEP_MW'].plot(label='Training Set')
df_test['AEP_MW'].plot(label='Test Set')
plt.axvline('2016-01-01', color='black', ls='--', lw=3)
plt.title('Data Splitting', weight='bold', fontsize=25)
plt.legend()
```
